# Découverte des listes

````python
# importation de la libraire random permettant de simuler l'aléatoire en python
import random


def randtab():
    tab = []
    for i in range(5):
        n  = random.randrange(10)
        tab.append(n)
    return tab
````

## Exercice (accumulateur)
Que fait la fonction ci-dessous ?
Que peut-on dire à tout instant de la variable acc (on parle d'invariant)

````python
def lala(tab):
    acc = 0
    for e in tab:
        print("acc : ", acc, "e : ", e)
        acc = acc + e
    return acc
````

````python
print(tab1, "lala(tab1) : ", lala(tab1))
print(tab2, "lala(tab2) : ", lala(tab2))
print(tab3, "lala(tab3) : ", lala(tab3))
print(tab4, "lala(tab4) : ", lala(tab4))
````

## Réponse
La fonction lala fait la somme des éléments dans le tableau.
La variable acc est à tout instant égale à la somme des éléments déjà parcourus par la boucle.

## Exercice (accumulateur)
Écrire une fonction prenant en paramètre une liste d'entier **tab** et retournant la moyenne des valeurs dans tab.

Commentez votre fonction et testez-la abondamment !

````python
def moyenne(tab):
    # Si la tab est vide la moyenne fait 0
    if not tab:
        return 0
    # On commence par faire la somme
    acc = 0
    for e in tab:
        print("acc : ", acc, "e : ", e)
        acc = acc + e
    # On divise par le nombre d'éléments
    return acc/len(tab)

print(tab1, "moyenne(tab1) : ", moyenne(tab1))
print(tab2, "moyenne(tab2) : ", moyenne(tab2))
print(tab3, "moyenne(tab3) : ", moyenne(tab3))
print(tab4, "moyenne(tab4) : ", moyenne(tab4))
````

## Exercice (recherche)
Programmer la fonction **recherche**, prenant en paramètre une liste non vide **tab**  d'entiers et un entier n, et qui renvoie True si n est dans tab et False sinon.

Commentez votre fonction et testez-la abondamment !

````python
def recherche(tab, n):
    # On parcours le tableau
    for e in tab:
        # On teste si on vient de trouver e
        if e == n:
            # Si on trouve on s'arrete et on retourne True
            return True
    # Si on a pas trouvé à la fin du tableau on retourne False
    return False

tab7 = [3, 5, 1, 4, 9, 6, 2, 5, 4, 3]
print(tab7)
if recherche(tab7, 0):
    print("0 a été trouvé")
else:
    print("0 n'a pas été trouvé")

if recherche(tab7, 2):
    print("2 a été trouvé")
else:
    print("2 n'a pas été trouvé")
````

````python
def recherche(tab, n):
    # On parcours le tableau
    for i, e in enumerate(tab):
        # On teste si on vient de trouver e
        if e == n:
            # Si on trouve on s'arrete et on retourne True
            return True, i
    # Si on a pas trouvé à la fin du tableau on retourne False
    return False, None

tab7 = [3, 5, 1, 4, 9, 6, 2, 5, 4, 3]
print(tab7)
trouve, indice = recherche(tab7, 0)
if trouve:
    print(f"0 a été trouvé a la position {indice}")
else:
    print("0 n'a pas été trouvé")
trouve, indice = recherche(tab7, 2)
if trouve:
    print(f"2 a été trouvé a la position {indice}")
else:
    print("2 n'a pas été trouvé")
````

## Exercice
Que fait la fonction ci-dessous ?

````python
def surprise(tab):
    longueur = len(tab)
    # Traverser tous les éléments de la liste
    print(tab)
    for i in range(longueur):
        print(f"{i} ème passe")
        # Pour tous les éléments de la liste sauf les i derniers
        for j in range(0, longueur-i-1):
            print(tab)
            # échanger si l'élément trouvé est plus grand que le suivant
            if tab[j] > tab[j+1] :
                permuter(tab, j, j+1)
    return tab
````

````python
testtab = randtab(20)
surprise(testtab)
````

## Réponse
La fonction ci-dessus tri le tableau, l'algorithme parcourt le tableau et compare les éléments consécutifs.
Lorsque deux éléments consécutifs ne sont pas dans l'ordre, ils sont permutés.

### Invariant :
Après un premier parcours complet du tableau, le plus grand élément est forcément en fin de tableau, à sa position définitive.
En effet, aussitôt que le plus grand élément est rencontré durant le parcours, il est mal trié par rapport à tous les éléments suivants, donc permuté avec le suivant jusqu'à arriver à la fin du parcours.

Après le premier parcours, le plus grand élément étant à sa position définitive, il n'a plus à être traité.
C'est pour cela qu'on ne s'intéresse plus qu'aux éléments le précédent.

Le reste du tableau est en revanche encore en désordre. Il faut donc le parcourir à nouveau, en s'arrêtant à l'avant-dernier élément.

Après ce deuxième parcours, les deux plus grands éléments sont à leur position définitive.
Il faut en conséquence répéter les parcours du tableau, jusqu'à ce que les deux plus petits éléments soient placés à leur position définitive.
On répète ainsi autant de fois que d'élément, mais toujours en parcourant un élément de moins.

## Exercice
Comment pourrait-on améliorer la fonction ci-dessus et pourquoi ?

````python
def tribulle(tab):
    longueur = len(tab)
    # Traverser tous les éléments de la liste
    print(tab)
    for i in range(longueur):
        permutation = 0
        print(f"{i} ème passe")
        # Pour tous les éléments de la liste sauf les i derniers
        for j in range(0, longueur-i-1):
            print(tab)
            # échanger si l'élément trouvé est plus grand que le suivant
            if tab[j] > tab[j+1] :
                permuter(tab, j, j+1)
                permutation = permutation +1
        print(f"{permutation} permutation")
        if not permutation :
            return tab
````

````python
tribulle(randtab(20))
````

## Réponse
On remarque que dès qu'une passe n'entraine aucune permutation, alors le tableau est trié.

Ainsi, à fin de gagner du temps, on peut décider d'arrêter la fonction dès qu'on observe ce comportement.

Pour cela, on introduit une variable permutation qui sert à compter le nombre de permutations lors de la passe. 
À la fin de chaque passe, on vérifie qu'au moins une permutation a été faite.
Dans le cas contraire, on arrête la fonction et on retourne le tableau.

## Réponse
Il existe différent algorithme de tri.
Les plus faciles à comprendre sont le tri par insertion et le tri par sélection.

Cette petite vidéo bien sympathique présente les tris les plus courants : [https://www.youtube.com/watch?v=ra79TDfotno](https://www.youtube.com/watch?v=ra79TDfotno)

Pour comprendre de manière fun avec des danses hongroises : [https://www.youtube.com/@megaovermoc](https://www.youtube.com/@megaovermoc)

Pour allez plus loin : [https://fr.wikipedia.org/wiki/Algorithme_de_tri](https://fr.wikipedia.org/wiki/Algorithme_de_tri)

````python
def triselection(tab):
    tab1 = []
    for _ in range(len(tab)):
        print(tab)
        current = tab[0]
        icurrent = 0
        # On cherche l'indice du min
        for i, e in enumerate(tab) :
            if e < current :
                current = e
                icurrent = i
        tab1.append(current)
        print("min :", current, "indice :", icurrent, "tab trie:", tab1)
        # On le supprime
        del tab[icurrent]
    tab = tab1
    return tab1
````

````python
triselection(randtab(20))
````
