# Découverte des listes

Pour rappel, une liste est une structure de données linéaire représentée par l'interface suivante :
- Longueur : retourne la longueur de la liste
- Ajouter : ajoute un élément dans la liste par son index
- Supprimer : supprime un élément de la liste par son index
- Accéder : accède à un élément de la liste par son index
- Liste vide : permet de tester si une liste est vide

En python, on utilisera la structure mutable list comme implémentation de la liste.

Dont voici la documentation officielle : [https://docs.python.org/fr/3/tutorial/datastructures.html](https://docs.python.org/fr/3/tutorial/datastructures.html)

Que fait la fonction ci-dessous ?


````python
# importation de la bibliothèque random permettant de simuler l'aléatoire en python
import random


def randtab():
    tab = []
    for i in range(5):
        n  = random.randrange(10)
        tab.append(n)
    return tab
````

````python
def surprise(tab):
    longueur = len(tab)
    # Traverser tous les éléments de la liste
    print(tab)
    for i in range(longueur):
        print(f"{i} ème passe")
        # Pour tous les éléments de la liste sauf les i derniers
        for j in range(0, longueur-i-1):
            print(tab)
            # échanger si l'élément trouvé est plus grand que le suivant
            if tab[j] > tab[j+1] :
                permuter(tab, j, j+1)
    return tab
````
