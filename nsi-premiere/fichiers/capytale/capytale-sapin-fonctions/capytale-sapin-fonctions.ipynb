!!! info Définir une fonction
Une fois **définie** la fonction `tronc`, on pourra, à tout moment redessiner ce tronc en **appelant** cette fonction : `tronc()`
!!!

!!! info Colorier une zone
 - `fillcolor("brown")` définit la **couleur de coloriage** du dessin en marron.

```python
def tronc():
    """ Dessine le tronc du sapin """
    fillcolor("brown")
    begin_fill()
    forward(5)
    right(90)
    forward(50)
    ...
    end_fill()
```
