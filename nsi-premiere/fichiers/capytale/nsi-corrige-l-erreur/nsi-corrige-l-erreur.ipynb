## 🧪 Niveau 2 : Données et Collections

### Question 3 : Calcul de la moyenne
Cette fonction doit calculer la moyenne d'une liste de notes.

```python
def calculer_moyenne(notes):
    total = 0
    for i in range(len(notes)):
        total = total + notes[i+1]
    return total / len(notes)

ma_classe = [12, 15, 10, 18]
print(calculer_moyenne(ma_classe))
```

??? tip "Indice sur les limites"
L'erreur est une `IndexError`. N'oubliez pas que dans une liste de taille $n$, les indices vont de **$0$ à $n-1$**. Si vous faites `i+1` à la fin de la boucle, vous sortez de la liste.
???

### Question 4 : Gestion du stock
On veut récupérer le prix d'un objet dans le dictionnaire du magasin.

```python
inventaire = {"épée": 50, "bouclier": 30, "potion": 10}

def prix_objet(nom_objet):
    prix = inventaire[nom_objet]
    return f"L'objet {nom_objet} coûte {prix} pièces."

print(prix_objet("Epee"))
```

??? tip "Indice sur les dictionnaires"
C'est une `KeyError`. Vérifiez l'orthographe exacte et les **accents**. Le dictionnaire et l'appel de la fonction doivent correspondre parfaitement.

### Question 5 : L'ajout de bonus
Cette fonction est censée ajouter un bonus à une liste de scores.

```python
def ajouter_bonus(liste_scores):
    nouvelle_liste = []
    for s in liste_scores:
        nouvelle_liste.append(s + 5)

mes_scores = [10, 20, 30]
mes_scores = ajouter_bonus(mes_scores)
mes_scores.append(40) # Doit ajouter 40 à la fin de la nouvelle liste
print(mes_scores)
```

??? tip "Indice sur le NoneType"
Si vous obtenez `AttributeError: 'NoneType' object has no attribute 'append'`, c'est que votre fonction a oublié de **renvoyer (`return`)** le résultat. Sans `return`, la variable devient vide (`None`).
???
