<h2 style="background-color: rgb(0,161,153); color: white; border-radius: 3px;padding: 10px;">Comment relier des machines entre-elles ?<br/>Internet : un réseau de réseaux</h2>

Que se passe-t-il lorsque vous tapez dans la barre d'adresse de votre navigateur http://www.wikipedia.fr ? 

Votre ordinateur va chercher à entrer en communication avec un autre ordinateur se trouvant probablement à des milliers de kilomètres de chez vous. Pour pouvoir établir cette communication, il faut bien sûr que les deux ordinateurs soient « reliés ». On dira que nos deux ordinateurs sont en réseau.

Il existe énormément de réseaux (la plupart des ordinateurs du lycée sont en « réseau »), certains réseaux sont reliés à d'autres réseaux qui sont eux-mêmes reliés à d'autres réseaux... ce qui forme « un réseaux de réseaux ». Savez-vous comment on appelle cet assemblage multiple de réseaux ? __Internet !__

![Internet](Internet.jpg)

__Internet est constitué de machines connectées par des câbles, ou parfois par ondes électromagnétiques (ex : WIFI) : c'est un réseau de réseaux__.

<h2 style="background-color: rgb(0,161,153); color: white; border-radius: 3px;padding: 10px;">Comment identifier une machine sur un réseau ?<br/>L'adresse IP</h2>

__L’adresse IP est une adresse unique attribuée à chaque machine sur un réseau__. Tout comme une adresse postale, il faut connaître au préalable l'adresse IP de l'ordinateur avec lequel on veut communiquer.

__L'adresse IP, en version IPv4, se présente sous forme de 4 octets__ séparés par des points. Par exemple : 204.35.129.3

On rappelle qu'un octet est composé de 8 bits (d'où son nom). Ci-dessous, un rappel de conversion d'un nombre binaire codé sur un octet (`10010110`) vers sa valeur décimale (`150`) :

<h2 style="background-color: rgb(0,161,153); color: white; border-radius: 3px;padding: 10px;">Comment numéroter toutes les machines du monde en IPv4 ?<br/>Les adresses IP privées et publiques</h2>

!!! info Définitions
- __Adresse IP publique :__ adresse numérique désignant une machine sur internet.
- __Adresse IP privée :__ adresse numérique désignant une machine sur un réseau local. Chez nous, elles commencent souvent par 192.168.1.X
- __Routeur :__ machine qui permet de transmettre des informations d’un réseau à un autre réseau.
!!!

Sachant que, __sans routeur, un ordinateur ne peut communiquer qu'avec les machines de son propre réseau__, on peut économiser énormément d'adresses IP publiques à l'aide des adresses privées :

- Dans chacun de nos réseaux locaux (à la maison, au lycée,...), on doit avoir des adresses IP privées différentes pour chaque machine du réseau mais... rien n'empêche d'avoir les mêmes adresses d'un réseau local à l'autre !
- On peut donc __utiliser les mêmes adresses dans tous les réseaux locaux du monde__.
- Par contre, il faut absolument __conserver des adresses IP publiques toutes différentes sur Internet__.

!!! info Le protocole IPv6
A terme, le vieillissant protocole IPv4 sera remplacé par l'__IPv6__, capable de nommer plus de __$10^{38}$ machines différentes__ !

Il ne sera plus nécessaire de différencier les adresses publiques et privées lorsque tous les réseaux d'Internet seront compatibles avec ce nouveau protocle.
!!!

<h2 style="background-color: rgb(0,161,153); color: white; border-radius: 3px;padding: 10px;">Comment savoir sur quel réseau une machine est connectée ?<br/>Les masques de sous-réseau</h2>

__Le masque de sous-réseau permet de savoir si plusieurs machines appartiennent au même réseau.__
Dans l’exemple ci-dessous, le __masque de sous-réseau 255.255.255.0__ indique que seul le dernier octet de l’adresse IP permet d’identifier une machine sur le __réseau 192.168.1.0__. 

![Masque de sous-réseau](masque.png)

__Les machines du réseau local seront donc de la forme 192.168.1.X__. Et donc, __on pourra ainsi former 254 adresses locales différentes__, donc connecter 254 appareils distincts dans le réseau. Dans l'exemple ci-dessus, 192.168.1.1 est une adresse IP de réseau local. 

![Réseau](Reseau.png)

!!! info Une adresse IP pour le routeur
Dans le réseau schématisé ci-dessus, on peut voir __un routeur, permettant de communiquer d'un réseau (local) à un autre (Internet)__.

> __Remarques :__
> - Chez nous, __on appelle le routeur une box__, c'est le boitier fourni par notre FAI (Orange, Free, SFR ou Bouygues).
> - Le dernier octet de l'adresse IP du routeur est souvent 254, par tradition et pour mieux le trouver.
!!!

| Machine | IP d'une machine | Masque de sous-réseau | Nom du réseau |
|:-----|:-----:|:-----:|:-----:|
|Ordinateur fixe|192.168.1.1|255.255.255.0|192.168.1.0|
|Ordinateur portable|192.168.1.2|255.255.255.0|192.168.1.0|
|Téléphone 1|192.168.1.45|255.255.255.0|192.168.1.0|
|Téléphone 2|192.168.1.167|255.255.255.0|192.168.1.0|
|Console de jeu|192.168.1.224|255.255.255.0|192.168.1.0|
|Routeur (Box)|192.168.1.254|255.255.255.0|192.168.1.0|

!!! tip 255 ou 0 ?
- Lorsqu'un octet du masque est égal à 255, cela siginifie que cet octet désigne le nom du réseau.
- Lorsqu'un octet du masque est égal à 0, cela siginifie que cet octet désigne le nom d'une machine sur le réseau.
!!!
